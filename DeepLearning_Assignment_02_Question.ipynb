## Write Answer 03:Activation functions are mathematical function apply to the output of a neuron after the linear transformation. They introduce non linearity into the neural network, enabling it to learn complex patterns from data.

# Why Do We Need Activation Functions :
Real-world data is usually non linear.
If we do not use activation functions the neural network will behave like a linear model.

Activation functions allow the network to learn:
1.Complex relationships
2.Non-linear decision boundaries
3.Real-world patterns
With activation functions, deep neural networks can approximate very complex functions.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
import numpy as np

class IQRHandler(BaseEstimator, TransformerMixin):

    def fit(self, X, y=None):
        self.Q1 = np.percentile(X, 25, axis=0)
        self.Q3 = np.percentile(X, 75, axis=0)
        self.IQR = self.Q3 - self.Q1
        return self

    def transform(self, X):
        lower = self.Q1 - 1.5 * self.IQR
        upper = self.Q3 + 1.5 * self.IQR
        return np.clip(X, lower, upper)
```

```python
numerical_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('outlier', IQRHandler()),
    ('scaler', StandardScaler())
])
```

```python
categorical_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore')),
])
```

```python
combine_pipe = ColumnTransformer(transformers=[
    ('num', numerical_pipe, numeric_features),
    ('cat', categorical_pipe, categorical_features)
]
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
class MySimpleNN():
  def __init__(self,X):
    self.weights = torch.randn(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias = torch.zeros(1,dtype=torch.float64,requires_grad=True)

  def forward(self,X):
    z = torch.matmul(X,self.weights)+self.bias
    y_pred = torch.sigmoid(z)
    return y_pred
  def loss_function(self,y_pred,y):
    #clamp
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred,epsilon,1-epsilon)
    # Calculate individual losses and then take the mean to get a scalar loss
    loss = -y*torch.log(y_pred)-(1-y)*torch.log(1-y_pred)
    return loss.mean()
```

```python
learning_rate = 0.1
epochs = 25
```

```python
model = MySimpleNN(X_train_tensor)

#define loop
for epoch in range(epochs):
  y_pred = model.forward(X_train_tensor)
  #loss calcultor
  # Convert y_train to a tensor and pass it as the target variable
  y_train_tensor = torch.tensor(y_train.values, dtype=torch.float64).reshape(-1, 1)
  loss = model.loss_function(y_pred, y_train_tensor)

  #backward
  loss.backward()

  #update
  with torch.no_grad(): # Perform updates without tracking gradients
    model.weights.data -= learning_rate * model.weights.grad
    model.bias.data -= learning_rate * model.bias.grad

  #zero gradient
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  #print loss in each epoch
  print(f"{epoch+1}, loss ={loss.item()}")
```

```python
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):
  def __init__(self,num_features):
    super(Model,self).__init__()
    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out
```

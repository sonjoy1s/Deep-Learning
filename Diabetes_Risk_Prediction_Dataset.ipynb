```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
import numpy as np

class IQRHandler(BaseEstimator, TransformerMixin):

    def fit(self, X, y=None):
        self.Q1 = np.percentile(X, 25, axis=0)
        self.Q3 = np.percentile(X, 75, axis=0)
        self.IQR = self.Q3 - self.Q1
        return self

    def transform(self, X):
        lower = self.Q1 - 1.5 * self.IQR
        upper = self.Q3 + 1.5 * self.IQR
        return np.clip(X, lower, upper)
```

```python
numerical_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('outlier', IQRHandler()),
    ('scaler', StandardScaler())
])
```

```python
categorical_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore')),
])
```

```python
combine_pipe = ColumnTransformer(transformers=[
    ('num', numerical_pipe, numeric_features),
    ('cat', categorical_pipe, categorical_features)
]
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train_transformed = combine_pipe.fit_transform(X_train)
```

```python
X_test_transformed = combine_pipe.transform(X_test)
```

```python
class MySimpleNN():
  def __init__(self,X):
    self.weights = torch.randn(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias = torch.zeros(1,dtype=torch.float64,requires_grad=True)

  def forward(self,X):
    z = torch.matmul(X,self.weights)+self.bias
    y_pred = torch.sigmoid(z)
    return y_pred
  def loss_function(self,y_pred,y):
    #clamp
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred,epsilon,1-epsilon)
    loss = -y*torch.log(y_pred)-(1-y)*torch.log(1-y_pred)
    return loss
```

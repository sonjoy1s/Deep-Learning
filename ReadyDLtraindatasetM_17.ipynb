```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

# Numpy Array to tensors

```python
#numpy to tensor
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)

#change dtype
X_train_tensor = torch.tensor(X_train_tensor, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_tensor, dtype=torch.float32)

#numpy to tensor
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)

y_train_tensor = torch.tensor(y_train_tensor,dtype= torch.float32)
y_test_tensor = torch.tensor(y_test_tensor, dtype=torch.float32)

# y_test_tensor = torch.from_numpy(y_test)
```

```python
type(X_train_tensor)
```

```python
import torch.nn as nn
class MySimpleNN(nn.Module):

  def __init__(self,num_features):
    super().__init__()

    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

```python
learning_rate = 0.1
loss_function = nn.BCELoss()
```

```python
model = MySimpleNN(X_train_tensor.shape[1])
optimizer = torch.optim.SGD(model.parameters(),lr=learning_rate)
#define loop
batch_size = 32
epochs = 25
n_sample = len(X_train_tensor)
for epoch in range(epochs):
   for start_index in range(0,n_sample,batch_size):
    end_index = start_index + batch_size
   # print(f"Start index {start_index} end index{end_index}")
    X_batch = X_train_tensor[start_index:end_index]
    y_batch = y_train_tensor[start_index:end_index]

    #forward pass
    y_pred = model(X_batch)

    loss = loss_function(y_pred,y_batch.reshape(-1,1))


    #update step
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()



#print step

   print(f" {epoch+1} , Loss = {loss.item()} ")


```

```python
# model evalution
model.eval()
with torch.no_grad():
  y_pred = model(X_test_tensor)
  y_pred = (y_pred > 0.5).float()
  accuracy = (y_pred.view(-1) == y_test_tensor).float().mean().item()
  print(f"Accuracy = {accuracy}")
```

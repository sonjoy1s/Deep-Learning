# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:

weight update formula : wnew = wold + n(y-y^)*x1
n = learning rate
y = actual/target
y^ = predict
xi = input
error = (y-y^)
now, wnew = wold + n(error)*xi

# Why update only when prediction is wrong?:

1. If the prediction is already correct, the perceptron doesn't need to change. weights are already good for that input.

2. Updating only on mistakes save computation and ensures learning focuses on errors, improving the model efficiently

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

## Write Answer 06:
The Perceptron Learning Rule is the method used to update the weight of a perceptron when it makes a wrong prediction.

formula : wnew = wold+n(y-y^)xi
where,
n = learning rate
y = true label
y^= predicted label
xi= input
(y-y^) = error

1.if the perceptron makes a mistake, adjust the weights to reduce the error.
2.if the perceptron is correct, no change is need.

# Why can adjusting one weight affect the classification of other points?:
1.The weight determine the slope and position of the decision boundary.

2.Changing a weight moves the line, which can:
*Fix the classification for the current wrong point.
*Also change which side other points fall on, possibly causing them to be missedclassified.
